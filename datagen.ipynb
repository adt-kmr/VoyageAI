```python
#Generate dataset for our project: Voyage AI, a data-driven travel analytics company, aims to understand user behaviour by grouping
#travellers into meaningful segments for personalized recommendations and targeted marketing. Using
#historical travel data such as trip frequency, average booking value, destination diversity, session
#duration, and search behaviour, develop a Python-based unsupervised machine learning clustering
#model. The project should include data cleaning, pre-processing, feature scaling, cluster evaluation,
#and interpretation of traveller segments to support data-driven business strategies.
#generate synthetic data for the project
import pandas as pd 
import numpy as np
from sklearn.datasets import make_blobs
# Set random seed for reproducibility
np.random.seed(42)
# Define parameters for synthetic data
n_samples =  100000
n_features = 5
n_clusters = 4
# Generate synthetic data using make_blobs
X, y = make_blobs(n_samples=n_samples, n_features=n_features, centers=n_clusters, cluster_std=1.0)      
# Create a DataFrame with meaningful column names
columns = ['trip_frequency', 'average_booking_value', 'destination_diversity', 'session_duration', 'search_behavior']
df = pd.DataFrame(X, columns=columns)
# Add some noise to simulate real-world data
df['trip_frequency'] = df['trip_frequency'] + np.random.normal(0,2,n_samples)
df['average_booking_value'] = df['average_booking_value'] + np.random.normal(0,500,n_samples)
df['destination_diversity'] = df['destination_diversity'] + np.random.normal(0,1,n_samples)
df['session_duration'] = df['session_duration'] + np.random.normal(0,10,n_samples)
df['search_behavior'] = df['search_behavior'] + np.random.normal(0,5,n_samples)
# Save the synthetic dataset to a CSV file
df.to_csv('dataai.csv', index=False)
print("Synthetic dataset 'voyage_ai_travel_data.csv' generated successfully.")
# to create a synthetic dataset for clustering analysis

```
