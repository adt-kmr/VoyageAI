```python
'''Generating a dataset of 100000rows for the project Voyage AI, a data-driven travel analytics company, aims to understand user behaviour by grouping
travellers into meaningful segments for personalized recommendations and targeted marketing. Using
historical travel data such as trip frequency, average booking value, destination diversity, session
duration, and search behaviour, develop a Python-based unsupervised machine learning clustering
model. The project should include data cleaning, pre-processing, feature scaling, cluster evaluation,
and interpretation of traveller segments to support data-driven business strategies
we have to generate a dataset of 100000 rows which is messy and include duplicate rows, missing values, and outliers and we have to show encoding for converting categorical cols into numerical columns. The dataset should have the following columns: 'user_id', 'trip_frequency', 'average_booking_value', 'destination_diversity', 'session_duration', and 'search_behavior'.'''
```

```python
# Generating a dataset of 100000 rows for the project
import pandas as pd
import numpy as np
import random
# Set random seed for reproducibility
random.seed(100000)
# Set the number of rows for the dataset
num_rows = 100000
# this is the list of search behaviour options
search_behaviors_list = ['frequentsearcher', 'occasionalsearcher', 'raresearcher']
# Generate random data for each column
user_ids = [i for i in range(1, num_rows + 1)]
trip_frequencies = np.random.poisson(lam=5, size=num_rows)
average_booking_values = np.random.normal(loc=500, scale=150, size=num_rows)
destination_diversities = np.random.randint(1, 10, size=num_rows)
session_durations = np.random.exponential(scale=30, size=num_rows)
search_behaviors = np.random.choice(search_behaviors_list, size=num_rows)
# Create a DataFrame
data = pd.DataFrame({
    'user_id': user_ids,
    'trip_frequency': trip_frequencies,
    'average_booking_value': average_booking_values,
    'destination_diversity': destination_diversities,
    'session_duration': session_durations,
    'search_behavior': search_behaviors
})
# Introduce some missing values randomly (5% per numeric column)
for col in ['trip_frequency', 'average_booking_value', 'destination_diversity', 'session_duration']:
    data.loc[data.sample(frac=0.05).index, col] = np.nan
# Introduce some duplicate rows randomly (2%)
duplicate_rows = data.sample(frac=0.02)
data = pd.concat([data, duplicate_rows], ignore_index=True)
# Introduce some outliers in 'average_booking_value' (make them extreme)
outliers = np.random.choice(data.index, size=100, replace=False)
data.loc[outliers, 'average_booking_value'] = data['average_booking_value'].mean() * 10
# Show the first few rows of the generated dataset
print(data.head())
# Show the shape of the dataset
print(data.shape)
# Show summary statistics of the dataset
print(data.describe())
# Show the number of missing values in each column
print(data.isnull().sum())
# Show the number of duplicate rows
print(f'Number of duplicate rows: {data.duplicated().sum()}')
# Encoding the 'search_behavior' column using one-hot encoding
encoded_data = pd.get_dummies(data, columns=['search_behavior'])
# Show the first few rows of the encoded dataset
print(encoded_data.head())
```

```python
# Save the dataset to a CSV file
data.to_csv('dataset.csv', index=False)
print("Dataset saved to 'dataset.csv'")

# Save the encoded dataset as well
encoded_data.to_csv('dataset_encoded.csv', index=False)
print("Encoded dataset saved to 'dataset_encoded.csv'")
```
