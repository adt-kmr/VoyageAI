# Voyage AI - Traveller Intelligence Analysis
## Visualizing Persona Segments and Marketing Strategy

This notebook provides a visual deep dive into the four primary traveler personas identified through our customer intelligence analysis. The goal is to translate raw metrics into actionable business insights.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set aesthetic style
sns.set_theme(style="whitegrid")
palette = ["#4C72B0", "#55A868", "#C44E52", "#8172B3"] # Muted professional colors
```

```python
# 1. Define Persona Data
personas = ["Jet-Setter", "Leisure Seeker", "Quiet Luxurist", "Budget Explorer"]
user_counts = [38308, 28740, 18965, 13989]
avg_booking = [550, 430, 700, 280]
session_duration = [65, 45, 85, 25]
dest_diversity = [12, 7, 18, 4]

df = pd.DataFrame({
    "Persona": personas,
    "Users": user_counts,
    "Avg_Booking_Value": avg_booking,
    "Session_Duration": session_duration,
    "Destination_Diversity": dest_diversity
})

display(df)
```

### 1. User Distribution by Persona
The Jet-Setter and Leisure Seeker segments represent the bulk of our user base.

```python
plt.figure(figsize=(10, 6))
plt.pie(df['Users'], labels=df['Persona'], autopct='%1.1f%%', startangle=140, colors=palette, explode=(0.05, 0, 0, 0))
plt.title("Customer Persona Distribution", fontsize=15, fontweight='bold')
plt.show()
```

### 2. Average Booking Value & Counterintuitive Insight
**Insight**: Note how the 'Quiet Luxurist' (Least frequent traveler) has the highest booking value, while the 'Budget Explorer' (often labeled as High Spenders in some contexts) has the lowest.

```python
plt.figure(figsize=(10, 6))
sns.barplot(x="Persona", y="Avg_Booking_Value", data=df, palette=palette)
plt.title("Average Booking Value ($) by Persona", fontsize=15, fontweight='bold')
plt.ylabel("Booking Value ($)")
plt.axhline(df['Avg_Booking_Value'].mean(), color='grey', linestyle='--', label='Average')
plt.legend()
plt.show()
```

### 3. Engagement & Complexity
Comparing session duration and destination diversity across segments.

```python
fig, ax1 = plt.subplots(figsize=(12, 6))

# Bar chart for Session Duration
ax1.bar(df['Persona'], df['Session_Duration'], alpha=0.6, color=palette, label='Session Duration (min)')
ax1.set_ylabel('Session Duration (min)', color='navy')
ax1.tick_params(axis='y', labelcolor='navy')

# Create a second axis for Destination Diversity
ax2 = ax1.twinx()
ax2.plot(df['Persona'], df['Destination_Diversity'], color='darkred', marker='o', linewidth=2, label='Dest Diversity')
ax2.set_ylabel('Unique Destinations Visited', color='darkred')
ax2.tick_params(axis='y', labelcolor='darkred')

plt.title("Engagement Metrics (Session Time vs Destination Diversity)", fontsize=15, fontweight='bold')
plt.show()
```
